The demo is fully reproducible but uses a reduced dataset, so results may differ from those reported in the full study.

# 1 SWOOP candidate generation (Pilot Experiment)

```python
PROJECT_ROOT = Path().resolve().parents[1]

DEMO_DIR = PROJECT_ROOT / "data" / "demo_data"
RAW_FILE = DEMO_DIR / "demo_node_train_valid_raw.csv"
SIG_FILE = DEMO_DIR / "demo_node_sigma0_v5.10.csv"
OCCU_FILE = DEMO_DIR / "demo_node_occu_mean.csv"
OUTPUT_FILE = DEMO_DIR / "demo_train_valid_candidates.csv"

print("Raw input:", RAW_FILE)
print("Sigma input:", SIG_FILE)
print("Occurrence input:", OCCU_FILE)
print("Output:", OUTPUT_FILE)
```

```python
## Load the demo raw dataset
df = pd.read_csv(RAW_FILE)
print(f"Loaded {len(df):,} raw rows")
df.head()
```

```python
# load demo rdr_sig0 and occurrence data
df_sig = pd.read_csv(SIG_FILE)
df_occu = pd.read_csv(OCCU_FILE)

print(f"Loaded {len(df_sig):,} sigma0 rows")
print(f"Loaded {len(df_occu):,} occurrence rows")

df_sig.head(), df_occu.head()
```

```python
# Standardize key columns
df = df.copy()
df_sig = df_sig.copy()
df_occu = df_occu.copy()

df["node_id"] = df["node_id"].astype(np.int64)

df_sig["node_id"] = df_sig["node_id"].astype(np.int64)
df_occu["node_id"] = df_occu["node_id"].astype(np.int64)

if "occurrence" in df_occu.columns:
    df_occu.loc[df_occu["occurrence"] == 255, "occurrence"] = 0

df["reach_id"] = df["node_id"] // 10000 * 10 + df["node_id"] % 10

print("Unique reaches:", df["reach_id"].nunique())
print("Unique nodes:", df["node_id"].nunique())
```

```python
def detect_step_change(
    denoised_signal,
    sig_signal,
    occu_signal,
    std_signal,
    width_signal,
    stability_scale=0.3
):
    denoised_signal = np.asarray(denoised_signal)
    sig_signal = np.asarray(sig_signal)
    occu_signal = np.asarray(occu_signal)
    std_signal = np.asarray(std_signal)
    width_signal = np.asarray(width_signal)

    differences = np.abs(np.diff(denoised_signal))
    step_points = np.where(differences > 0.5)[0] + 1

    window_size1_default = 10
    window_size2_default = 10
    results = []

    if len(step_points) == 0:
        return None, False

    signal_range = np.max(denoised_signal) - np.min(denoised_signal)
    if signal_range == 0:
        return None, False

    n = len(denoised_signal)

    min_gap_nodes = 2
    min_sep = min_gap_nodes + 1

    step_points_sorted = np.sort(step_points)
    clusters = []
    current_cluster = [step_points_sorted[0]]

    for p in step_points_sorted[1:]:
        if p - current_cluster[-1] <= min_sep:
            current_cluster.append(p)
        else:
            clusters.append(current_cluster)
            current_cluster = [p]
    clusters.append(current_cluster)

    filtered_step_points = []
    for cluster in clusters:
        cluster = np.asarray(cluster)
        best_point = cluster[np.argmax(differences[cluster - 1])]
        filtered_step_points.append(int(best_point))

    for step_index in filtered_step_points:
        w1 = window_size1_default
        w2 = window_size2_default

        if step_index < w1 or step_index >= n - w2:
            w1 = min(w1, step_index)
            w2 = min(w2, n - step_index)

        if w1 <= 4 or w2 <= 5:
            continue

        before_segment = denoised_signal[step_index - w1: step_index]
        after_segment = denoised_signal[step_index + 1: step_index + w2]

        change_size1 = differences[step_index - 1]

        before_x = np.arange(len(before_segment))
        after_x = np.arange(len(after_segment))

        try:
            before_trend_coeff = np.polyfit(before_x, before_segment, 1)
            after_trend_coeff = np.polyfit(after_x, after_segment, 1)
        except Exception:
            continue

        before_fit = np.polyval(before_trend_coeff, before_x)
        after_fit = np.polyval(after_trend_coeff, after_x)

        before_residuals = before_segment - before_fit
        after_residuals = after_segment - after_fit

        before_residual_std = np.std(before_residuals)
        after_residual_std = np.std(after_residuals)

        before_extrap = np.polyval(before_trend_coeff, len(before_segment))
        after_extrap = np.polyval(after_trend_coeff, -1)

        change_size2 = abs(before_extrap - after_extrap)
        theta = np.arctan(before_trend_coeff[0])
        change_size2 = change_size2 * np.cos(theta)
        change_size3 = (before_extrap - after_extrap) * np.cos(theta)

        change_size = min(change_size1, change_size2)
        stability_threshold = stability_scale * change_size

        l = max(0, step_index - 5)
        r = min(n, step_index + 5)

        left_idx = np.arange(l, step_index)
        right_idx = np.arange(step_index + 1, r)
        idx = np.concatenate([left_idx, right_idx])

        if idx.size == 0:
            continue

        sig = np.nanmean(sig_signal[idx])
        sig2 = np.nanmean(sig_signal[left_idx]) if left_idx.size > 0 else np.nan
        sig3 = np.nanmean(sig_signal[right_idx]) if right_idx.size > 0 else np.nan
        sig4 = sig_signal[step_index]

        occu = np.nanmean(occu_signal[idx])
        occu2 = np.nanmean(occu_signal[left_idx]) if left_idx.size > 0 else np.nan
        occu3 = np.nanmean(occu_signal[right_idx]) if right_idx.size > 0 else np.nan
        occu4 = occu_signal[step_index]

        time_std = np.mean(std_signal[idx]) / signal_range

        width_vals = width_signal[idx]
        width_mean = np.mean(width_vals)
        width_std = np.std(width_vals)

        if (
            before_residual_std < stability_threshold
            and after_residual_std < stability_threshold
        ):
            results.append([
                step_index,
                change_size,
                change_size1,
                change_size2,
                change_size3,
                signal_range,
                sig,
                sig2,
                sig3,
                sig4,
                occu,
                occu2,
                occu3,
                occu4,
                time_std,
                before_residual_std,
                after_residual_std,
                before_trend_coeff[0],
                after_trend_coeff[0],
                width_mean,
                width_std,
            ])

    if len(results) > 0:
        return results, True

    return None, False
```

```python
median_df = (
    df.groupby("node_id")["wse"]
    .median()
    .reset_index(name="wse_median")
)

std_df = (
    df.groupby("node_id")["wse"]
    .std()
    .reset_index(name="wse_std")
)

width_df = (
    df.groupby("node_id")["width"]
    .mean()
    .reset_index(name="width")
)

average_df = median_df.merge(std_df, on="node_id", how="left")
average_df = average_df.merge(width_df, on="node_id", how="left")

print(f"Prepared node-level summary statistics for {len(average_df):,} nodes")
average_df.head()
```

```python
average_df = average_df.merge(df_sig, on="node_id", how="left")
average_df = average_df.merge(df_occu, on="node_id", how="left")

average_df["reach_id"] = (
    average_df["node_id"] // 10000 * 10 + average_df["node_id"] % 10
)

average_df = average_df.sort_values(["reach_id", "node_id"]).reset_index(drop=True)

print("Merged node-level table shape:", average_df.shape)
average_df.head()
```

```python
# Detect candidate step changes for each reach
results = []

for reach_id, group in average_df.groupby("reach_id"):
    group = group.sort_values("node_id").reset_index(drop=True)

    signal = group["wse_median"].values

    infos_list, has_step = detect_step_change(
        signal,
        group["rdr_sig0"].values,
        group["occurrence"].values,
        group["wse_std"].values,
        group["width"].values,
        stability_scale=0.3,
    )

    if has_step:
        for infos in infos_list:
            step_idx = infos[0]

            results.append({
                "reach_id": reach_id,
                "node_id": group["node_id"].iloc[step_idx],
                "step_index": step_idx,
                "change_size": infos[1],
                "change_size1": infos[2],
                "change_size2": infos[3],
                "change_size3": infos[4],
                "change_ratio": infos[1] / infos[5] if infos[5] != 0 else np.nan,
                "sigma": infos[6],
                "sigma2": infos[7],
                "sigma3": infos[8],
                "sigma4": infos[9],
                "occu": infos[10],
                "occu2": infos[11],
                "occu3": infos[12],
                "occu4": infos[13],
                "time_std": infos[14],
                "before_std": infos[15],
                "after_std": infos[16],
                "before_slope": infos[17],
                "after_slope": infos[18],
                "width_mean": infos[19],
                "width_std": infos[20],
            })

step_df = pd.DataFrame(results)

print(f"Detected {len(step_df):,} candidate steps")
step_df.head()
```

```python
print("Candidate reaches:", step_df["reach_id"].nunique() if len(step_df) > 0 else 0)
print("Candidate nodes:", step_df["node_id"].nunique() if len(step_df) > 0 else 0)

step_df.describe(include="all")
```

```python
step_df.to_csv(OUTPUT_FILE, index=False)

print(f"Saved detection results to: {OUTPUT_FILE}")
print(f"Number of detected candidate steps: {len(step_df)}")
```

# 2 Prepare CC-based features for demo candidate nodes

```python
# Input / output files for step 2
CANDIDATE_FILE = DEMO_DIR / "demo_train_valid_candidates_annotated.csv"
CC_OUTPUT_FILE = DEMO_DIR / "demo_train_valid_candidates_annotated_withcc.csv"

print("Candidate input:", CANDIDATE_FILE)
print("CC feature output:", CC_OUTPUT_FILE)
```

```python
def _window_corr_matrix(Y, X, left, right):
    """
    Compute window-based Pearson correlations between daily WSE profiles
    and the median WSE profile for all node positions within a reach.
    """
    D, N = Y.shape
    X2 = X * X

    # Both daily and median values must be finite
    M = np.isfinite(Y) & np.isfinite(X)[None, :]
    Mf = M.astype(float)

    # Replace invalid entries with zero for cumulative sums
    Y0 = np.where(M, Y, 0.0)
    Y20 = np.where(M, Y * Y, 0.0)
    X0 = X[None, :] * Mf
    X20 = X2[None, :] * Mf
    XY0 = X[None, :] * Y0

    def cumsum_pad(A):
        cs = np.cumsum(A, axis=1)
        return np.concatenate([np.zeros((D, 1), dtype=float), cs], axis=1)

    csY = cumsum_pad(Y0)
    csY2 = cumsum_pad(Y20)
    csX = cumsum_pad(X0)
    csX2 = cumsum_pad(X20)
    csXY = cumsum_pad(XY0)
    csN = cumsum_pad(Mf)

    sumY = csY[:, right] - csY[:, left]
    sumY2 = csY2[:, right] - csY2[:, left]
    sumX = csX[:, right] - csX[:, left]
    sumX2 = csX2[:, right] - csX2[:, left]
    sumXY = csXY[:, right] - csXY[:, left]
    nwin = csN[:, right] - csN[:, left]

    num = nwin * sumXY - sumX * sumY
    denx = nwin * sumX2 - sumX * sumX
    deny = nwin * sumY2 - sumY * sumY
    den = np.sqrt(denx * deny)

    corr = np.full((D, N), np.nan, dtype=float)
    ok = (nwin >= 2) & (den > 0)
    corr[ok] = num[ok] / den[ok]

    return corr, nwin
```

```python
def compute_node_profile_cc_fast(df, df_ref, window=5, ref_node_col="node_id"):
    """
    Compute local profile-correlation features for selected reference nodes.

    Parameters
    ----------
    df : pd.DataFrame
        Must contain at least ['reach_id', 'node_id', 'date', 'wse'].
    df_ref : pd.DataFrame
        Candidate table containing target node list.
    window : int
        Half window size along the reach.
    ref_node_col : str
        Column name in df_ref that stores node_id.

    Returns
    -------
    pd.DataFrame
        Node-level CC feature table.
    """
    df = df.copy()
    df["node_id"] = df["node_id"].astype("int64")
    df["date"] = pd.to_datetime(df["date"])

    ref_nodes = df_ref[ref_node_col].astype("int64").unique()

    # Median WSE profile for each node
    median_prof = (
        df.groupby(["reach_id", "node_id"], as_index=False)["wse"]
        .median()
        .rename(columns={"wse": "wse_median"})
    )

    # Daily WSE profile (median per node per day)
    daily_prof = (
        df.groupby(["reach_id", "date", "node_id"], as_index=False)["wse"]
        .median()
        .rename(columns={"wse": "wse_daily"})
    )

    # Derive reach_id for reference nodes
    ref_df = pd.DataFrame({"node_id": ref_nodes})
    ref_df["reach_id"] = ref_df["node_id"] // 10000 * 10 + ref_df["node_id"] % 10

    results = []

    for reach_id, gref in ref_df.groupby("reach_id"):
        nodes = (
            median_prof.loc[median_prof["reach_id"] == reach_id, "node_id"]
            .astype("int64")
            .unique()
        )

        if len(nodes) == 0:
            for nid in gref["node_id"].tolist():
                results.append(
                    dict(
                        node_id=np.int64(nid),
                        reach_id=reach_id,
                        cc_mean=np.nan,
                        cc_median=np.nan,
                        cc_count=0,
                        valid_nodes_mean=np.nan,
                        valid_nodes_median=np.nan,
                        window_size=0,
                    )
                )
            continue

        nodes = np.sort(nodes)
        N = len(nodes)

        # Median profile vector
        X = (
            median_prof[median_prof["reach_id"] == reach_id]
            .set_index("node_id")
            .reindex(nodes)["wse_median"]
            .to_numpy(dtype=float)
        )

        # Daily profile matrix
        sub_daily = daily_prof[daily_prof["reach_id"] == reach_id]
        if sub_daily.empty:
            for nid in gref["node_id"].tolist():
                results.append(
                    dict(
                        node_id=np.int64(nid),
                        reach_id=reach_id,
                        cc_mean=np.nan,
                        cc_median=np.nan,
                        cc_count=0,
                        valid_nodes_mean=np.nan,
                        valid_nodes_median=np.nan,
                        window_size=0,
                    )
                )
            continue

        Ydf = (
            sub_daily.pivot(index="date", columns="node_id", values="wse_daily")
            .reindex(columns=nodes)
        )
        Y = Ydf.to_numpy(dtype=float)

        pos = np.arange(N)
        left = np.maximum(0, pos - window)
        right = np.minimum(N, pos + window + 1)

        corr, nwin = _window_corr_matrix(Y, X, left, right)

        node_to_pos = {np.int64(n): i for i, n in enumerate(nodes)}

        for nid in gref["node_id"].tolist():
            nid = np.int64(nid)

            if nid not in node_to_pos:
                results.append(
                    dict(
                        node_id=nid,
                        reach_id=reach_id,
                        cc_mean=np.nan,
                        cc_median=np.nan,
                        cc_count=0,
                        valid_nodes_mean=np.nan,
                        valid_nodes_median=np.nan,
                        window_size=0,
                    )
                )
                continue

            p = node_to_pos[nid]
            s = corr[:, p]
            nn = nwin[:, p]

            cc_count = int(np.isfinite(s).sum())
            cc_mean = float(np.nanmean(s)) if cc_count else np.nan
            cc_median = float(np.nanmedian(s)) if cc_count else np.nan

            vn_count = int(np.isfinite(nn).sum())
            vn_mean = float(np.nanmean(nn)) if vn_count else np.nan
            vn_median = float(np.nanmedian(nn)) if vn_count else np.nan

            win_size = int(right[p] - left[p])

            results.append(
                dict(
                    node_id=nid,
                    reach_id=reach_id,
                    cc_mean=cc_mean,
                    cc_median=cc_median,
                    cc_count=cc_count,
                    valid_nodes_mean=vn_mean,
                    valid_nodes_median=vn_median,
                    window_size=win_size,
                )
            )

    out = (
        pd.DataFrame(results)
        .sort_values(["reach_id", "node_id"])
        .reset_index(drop=True)
    )

    return out
```

```python
# Load demo raw data again for CC computation
df_demo = pd.read_csv(RAW_FILE)
df_ref = pd.read_csv(CANDIDATE_FILE)

print(f"Loaded demo raw rows: {len(df_demo):,}")
print(f"Loaded candidate rows: {len(df_ref):,}")

print(df_ref.columns)
```

```python
# Prepare date / reach_id fields
df_demo = df_demo.copy()
df_demo["node_id"] = df_demo["node_id"].astype(np.int64)
df_demo["reach_id"] = df_demo["node_id"] // 10000 * 10 + df_demo["node_id"] % 10

# Same logic as 01.ipynb: convert SWOT-style time to calendar date
# If your demo raw file already has a usable date column, replace this block accordingly.
if "time" in df_demo.columns:
    df_demo["time_str"] = pd.to_datetime(df_demo["time"], unit="s", origin="2000-01-01")
    df_demo["date"] = pd.to_datetime(df_demo["time_str"].dt.date)
elif "date" in df_demo.columns:
    df_demo["date"] = pd.to_datetime(df_demo["date"])
else:
    raise ValueError("RAW_FILE must contain either 'time' or 'date' column for CC calculation.")

print(df_demo[["node_id", "reach_id", "date", "wse"]].head())
```

```python
# Compute CC features for candidate nodes
node_cc = compute_node_profile_cc_fast(
    df=df_demo[["reach_id", "node_id", "date", "wse"]],
    df_ref=df_ref,
    window=5,
    ref_node_col="node_id",
)

print(f"Computed CC features for {len(node_cc):,} candidate nodes")
node_cc.head()
```

```python
# Merge CC features back to candidate table
df_candidates_cc = df_ref.merge(
    node_cc[
        [
            "node_id",
            "cc_mean",
            "cc_median",
            "cc_count",
            "valid_nodes_mean",
            "valid_nodes_median",
            "window_size",
        ]
    ],
    on="node_id",
    how="left",
)

print("Merged candidate table shape:", df_candidates_cc.shape)
df_candidates_cc.head()
```

```python
# Save output for downstream RF construction
df_candidates_cc.to_csv(CC_OUTPUT_FILE, index=False)

print(f"Saved CC-enhanced candidate table to: {CC_OUTPUT_FILE}")
```

```python
test_average_df = test_average_df.merge(df_test_sig, on="node_id", how="left")
test_average_df = test_average_df.merge(df_test_occu, on="node_id", how="left")

test_average_df["reach_id"] = (
    test_average_df["node_id"] // 10000 * 10 + test_average_df["node_id"] % 10
)

test_average_df = test_average_df.sort_values(["reach_id", "node_id"]).reset_index(drop=True)

print("Merged test node-level table shape:", test_average_df.shape)
test_average_df.head()
```

```python
test_step_df.to_csv(TEST_STEP_FILE, index=False)

print(f"Saved test step candidates to: {TEST_STEP_FILE}")
print(f"Candidate reaches: {test_step_df['reach_id'].nunique() if len(test_step_df) > 0 else 0}")
print(f"Candidate nodes: {test_step_df['node_id'].nunique() if len(test_step_df) > 0 else 0}")
```

```python
if len(test_step_df) == 0:
    test_step_cc_df = test_step_df.copy()
    test_step_cc_df["cc_mean"] = np.nan
    test_step_cc_df["cc_median"] = np.nan
    test_step_cc_df["cc_count"] = 0
    test_step_cc_df["valid_nodes_mean"] = np.nan
    test_step_cc_df["valid_nodes_median"] = np.nan
    test_step_cc_df["window_size"] = 0
else:
    test_node_cc = compute_node_profile_cc_fast(
        df=df_test_cc[["reach_id", "node_id", "date", "wse"]],
        df_ref=test_step_df,
        window=5,
        ref_node_col="node_id",
    )

    test_step_cc_df = test_step_df.merge(
        test_node_cc[
            [
                "node_id",
                "cc_mean",
                "cc_median",
                "cc_count",
                "valid_nodes_mean",
                "valid_nodes_median",
                "window_size",
            ]
        ],
        on="node_id",
        how="left",
    )

print("Test candidate table with CC shape:", test_step_cc_df.shape)
test_step_cc_df.head()
```

```python
test_step_cc_df.to_csv(TEST_STEP_CC_FILE, index=False)

print(f"Saved CC-enhanced test candidate table to: {TEST_STEP_CC_FILE}")
print(f"Number of final test candidates: {len(test_step_cc_df)}")
```

# 4 Train RF on demo annotated candidates and apply to demo test candidates

```python
from sklearn.ensemble import RandomForestClassifier

# Best RF parameters copied from Experiment-15.ipynb
BEST_RF_PARAMS = {
    "n_estimators": 500,
    "max_depth": 8,
    "min_samples_leaf": 10,
    "max_features": "sqrt",
}

PRED_THRESHOLD = 0.5

BEST_RF_PARAMS
```

```python
TRAIN_ANNOTATED_FILE = DEMO_DIR / "demo_train_valid_candidates_annotated_withcc.csv"
TEST_FEATURE_FILE = DEMO_DIR / "demo_eu_test_step_candidates_with_cc.csv"

TEST_RF_ALL_FILE = DEMO_DIR / "demo_eu_test_rf_predictions.csv"
TEST_RF_POS_FILE = DEMO_DIR / "demo_eu_test_rf_positive.csv"

print("Annotated training file:", TRAIN_ANNOTATED_FILE)
print("Test feature file:", TEST_FEATURE_FILE)
print("RF all-prediction output:", TEST_RF_ALL_FILE)
print("RF positive output:", TEST_RF_POS_FILE)
```

```python
df_train = pd.read_csv(TRAIN_ANNOTATED_FILE)

# Keep only manually labeled samples
df_train = df_train.dropna(subset=["T/F"]).copy()

# Convert labels to binary
df_train["label"] = (df_train["T/F"] == "Y").astype(int)

print("Training samples:", len(df_train))
print("Positive samples:", int(df_train["label"].sum()))
print("Negative samples:", int((df_train["label"] == 0).sum()))

df_train.head()
```

```python
feat_cols = [
    "change_size1",
    "change_size2",
    "change_size3",
    "change_size",
    "sigma",
    "sigma2",
    "sigma3",
    "sigma4",
    "occu",
    "occu2",
    "occu3",
    "occu4",
    "time_std",
    "before_std",
    "after_std",
    "before_slope",
    "after_slope",
    "width_mean",
    "cc_mean",
]
```

```python
X = df_train[feat_cols].copy()
X = pd.get_dummies(X, dummy_na=True)
y = df_train["label"].values

print("Feature matrix shape:", X.shape)
print("Feature columns:")
print(list(X.columns))
```

```python
rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    class_weight="balanced",
    **BEST_RF_PARAMS,
)

rf.fit(X, y)

print("Final RF trained on full demo annotated data.")
print("Training rows:", X.shape[0])
print("Training columns:", X.shape[1])
```

```python
feat_importance = pd.DataFrame({
    "feature": X.columns,
    "importance": rf.feature_importances_
}).sort_values("importance", ascending=False).reset_index(drop=True)

feat_importance.head(20)
```

```python
df_test_rf = pd.read_csv(TEST_FEATURE_FILE)

print("Total test candidates:", len(df_test_rf))
df_test_rf.head()
```

```python
X_test = df_test_rf[feat_cols].copy()
X_test = pd.get_dummies(X_test, dummy_na=True)

# Align test columns to training columns
X_test = X_test.reindex(columns=X.columns, fill_value=0)

print("Test feature matrix shape:", X_test.shape)
```

```python
y_prob_test = rf.predict_proba(X_test)[:, 1]
y_pred_test = (y_prob_test >= PRED_THRESHOLD).astype(int)
y_label_test = np.where(y_pred_test == 1, "Y", "N")

df_test_rf["pred_prob"] = y_prob_test
df_test_rf["pred"] = y_pred_test
df_test_rf["pred_label"] = y_label_test

print("Predicted positive test candidates:", int((df_test_rf["pred"] == 1).sum()))
print("Predicted positive ratio:", float((df_test_rf["pred"] == 1).mean()))
df_test_rf.head()
```

```python
df_test_rf.to_csv(TEST_RF_ALL_FILE, index=False)

df_test_rf_pos = df_test_rf[df_test_rf["pred"] == 1].copy()
df_test_rf_pos.to_csv(TEST_RF_POS_FILE, index=False)

print(f"Saved all predictions to {TEST_RF_ALL_FILE}")
print(f"Saved predicted positive candidates to {TEST_RF_POS_FILE}")
print("Positive subset rows:", len(df_test_rf_pos))
```
